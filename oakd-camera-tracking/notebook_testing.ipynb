```python
from pathlib import Path

# Get the current notebook directory
script_dir = Path.cwd()  # Current working directory
examples_root = (
    script_dir / Path("../oakd-camera-tracking")
).resolve()  # This resolves the parent directory correctly
models = examples_root / "src" / "models"

print(f"Script Dir: {script_dir}")
print(f"Examples Root: {examples_root}")
print(f"Models Dir: {models}")
```

```python
import depthai as dai

# Create pipeline
with dai.Pipeline() as pipeline:
    # Test different camera sockets
    camera_sockets = [
        dai.CameraBoardSocket.CAM_A,  # RGB camera
        dai.CameraBoardSocket.CAM_B,  # Left mono camera
        dai.CameraBoardSocket.CAM_C,  # Right mono camera
    ]

    socket_names = {
        dai.CameraBoardSocket.CAM_A: "CAM_A (RGB)",
        dai.CameraBoardSocket.CAM_B: "CAM_B (Left Mono)",
        dai.CameraBoardSocket.CAM_C: "CAM_C (Right Mono)",
    }

    print("OAK-D Lite Camera Output Sizes:\n")

    for socket in camera_sockets:
        try:
            cam = pipeline.create(dai.node.Camera).build(socket)

            # Get sensor information
            sensor_resolution = cam.getResolutionSize()
            video_size = cam.getVideoSize()
            preview_size = cam.getPreviewSize()
            still_size = cam.getStillSize()

            print(f"{socket_names[socket]}:")
            print(f"  Sensor Resolution: {sensor_resolution}")
            print(f"  Video Size: {video_size}")
            print(f"  Preview Size: {preview_size}")
            print(f"  Still Size: {still_size}")
            print()

        except Exception as e:
            print(f"{socket_names[socket]}: Error - {e}\n")
```
