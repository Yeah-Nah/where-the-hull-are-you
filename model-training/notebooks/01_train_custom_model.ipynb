## Setup and Imports

```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent / "src"))

from training import ModelTrainer
from training.config import TrainingConfig
```

## Load Configuration

```python
config_path = Path("../config/training_config.yaml")
config = TrainingConfig(config_path)

print("Training configuration loaded:")
print(f"  Base model: {config.get('model.base_model')}")
print(f"  Epochs: {config.get('training.epochs')}")
print(f"  Batch size: {config.get('training.batch_size')}")
```

## Initialize Trainer

```python
trainer = ModelTrainer(config.config)
trainer.load_base_model(config.get("model.base_model"))
```

## Train Model

```python
data_yaml = Path(config.get("data.dataset_yaml"))

best_model_path = trainer.train(
    data_yaml=data_yaml,
    epochs=config.get("training.epochs"),
    imgsz=config.get("training.image_size"),
    batch=config.get("training.batch_size"),
    project=config.get("output.project"),
    name=config.get("output.name"),
)

print("\nTraining complete!")
print(f"Best model saved to: {best_model_path}")
```

## Validate Model

```python
validation_metrics = trainer.validate(data_yaml)

print("\nValidation Results:")
for metric_name, value in validation_metrics.items():
    print(f"  {metric_name}: {value:.4f}")
```

## Save Model to Custom Directory

```python
import shutil

save_dir = Path(config.get("output.save_dir"))
save_dir.mkdir(parents=True, exist_ok=True)

# Copy best weights
final_path = save_dir / "best.pt"
shutil.copy(best_model_path, final_path)

print(f"\nModel saved to: {final_path}")
```
